```python
SUMMARY_STAT_NAMES = [
    'tumor_cell_count',         # total tumor cells at T2
    'cd8_exhausted_fraction',   # fraction of CD8s that are terminally exhausted
    'cd8_active_fraction',      # fraction of CD8s that are active/non-exhausted
    'treg_fraction',            # Tregs as fraction of total CD4
    'm1_m2_ratio',              # M1 / (M1 + M2) macrophage ratio
    'mdsc_count',               # total MDSC count
    'tumor_immune_ratio',       # tumor cells / total immune cells (tumor burden proxy)
    'pdl1_expression',     # mean PD-L1 on tumor cells
]
```

```python
mean_paths = []
for summary in SUMMARY_STAT_NAMES:
    value = f'{summary}_sd_sobol.csv'
    path = '/athena/cayuga_0014/scratch/kch4018/abm-files/TME_Simulation/ABM_Jul25/sensitivity_results'
    path = f'{path}/{value}'
    mean_paths.append(path)
all_dfs = {
    summary: pd.read_csv(path)
    for summary, path in zip(SUMMARY_STAT_NAMES, mean_paths)
}

for summary, df in all_dfs.items():
    # Make all S1 and ST values between 0 and 1
    df["S1"] = df["S1"].clip(0, 1)
    df["ST"] = df["ST"].clip(0, 1)
    print(f"\n{summary}")
    print(
        df.nlargest(5, "ST")[
            ["parameter", "ST", "relative_ST_conf"]
        ]
    )
```

```python
df = pd.read_csv('/athena/cayuga_0014/scratch/kch4018/abm-files/TME_Simulation/ABM_Jul25/sensitivity_results/all_runs.csv')
for stat in SUMMARY_STAT_NAMES:
    cv = df[f"{stat}_sd"] / df[f"{stat}_mean"]
    mean = df[f"{stat}_mean"]
    sd = df[f"{stat}_sd"]
    print(stat)
    print(cv.describe())
    print(mean.describe())
    print(sd.describe())
```

```python
for stat in [
    "tumor_cell_count",
    "cd8_exhausted_fraction",
    "cd8_active_fraction",
    "treg_fraction",
    "m1_m2_ratio",
    "mdsc_count",
    "tumor_immune_ratio",
    "pdl1_expression"
]:
    sd_values = df[f"{stat}_sd"]
    print(stat)
    print(sd_values.describe())
```
